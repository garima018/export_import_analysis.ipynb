```python
year_export = export_data.groupby("year")["value"].sum()
year_import = import_data.groupby("year")["value"].sum()

plt.plot(year_export.index, year_export.values, label="Export")
plt.plot(year_import.index, year_import.values, label="Import")

plt.title("Export vs Import Trend")
plt.xlabel("Year")
plt.ylabel("Total Value")
plt.legend()
plt.show()
```

```python
top_export = export_data.groupby("Commodity")["value"].sum().nlargest(10)

labels = [x[:18] + "..." if len(x) > 18 else x for x in top_export.index]

plt.figure(figsize=(10, 6))
plt.bar(labels, top_export.values)

plt.title("Top 10 Export Commodities")
plt.xlabel("Commodity")
plt.ylabel("Export Value")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```

```python
top_import = import_data.groupby("Commodity")["value"].sum().nlargest(10)
labels = [x[:15] + "..." if len(x) > 15 else x for x in top_import.index]

plt.figure(figsize=(10,6))
plt.bar(labels, top_import.values)
plt.title("Top 10 Import Commodities")
plt.xlabel("Commodity")
plt.ylabel("Import Value")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.boxplot(export_data["value"].dropna())

plt.title("Export Value Distribution")
plt.ylabel("Export Value")
plt.xticks([1], ["Export Value"])

plt.tight_layout()
plt.show()
```

```python
e = export_data.groupby("year")["value"].sum()
i = import_data.groupby("year")["value"].sum()

common = e.index.intersection(i.index)

plt.figure(figsize=(9,6))
plt.scatter(e.loc[common], i.loc[common])
plt.title("Export vs Import Relationship")
plt.xlabel("Export Value")
plt.ylabel("Import Value")
plt.tight_layout()
plt.show()
```

```python
top_country = export_data.groupby("country")["value"].sum().nlargest(5)

country_names = {
    "U Arab Emts": "United Arab Emirates",
    "China P Rp": "China"
}

labels = [country_names.get(x, x) for x in top_country.index]

plt.figure(figsize=(8, 8))

plt.pie(
    top_country.values,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Top 5 Export Countries")
plt.tight_layout()
plt.show()
```

```python
print("KEY INSIGHTS")
print("--------------------")

print("1. Total Export Value:", export_data["value"].sum())
print("2. Total Import Value:", import_data["value"].sum())

print("3. Top Export Commodity:",
      export_data.groupby("Commodity")["value"].sum().idxmax())

print("4. Top Import Commodity:",
      import_data.groupby("Commodity")["value"].sum().idxmax())

print("5. Top Export Country:",
      export_data.groupby("country")["value"].sum().idxmax())
```

```python
export_country = export_data.groupby("country")["value"].sum()
import_country = import_data.groupby("country")["value"].sum()

country_trade = pd.DataFrame({
    "Export": export_country,
    "Import": import_country
}).fillna(0)

country_trade["Total Trade"] = country_trade["Export"] + country_trade["Import"]

country_trade.sort_values("Total Trade", ascending=False).head(10)
```

```python
country_trade["Trade Balance"] = country_trade["Export"] - country_trade["Import"]

country_trade.sort_values("Trade Balance", ascending=False).head(10)
```

```python
year_export = export_data.groupby("year")["value"].sum()
year_import = import_data.groupby("year")["value"].sum()

plt.figure(figsize=(9, 6))

plt.plot(year_export.index, year_export.values, marker="o", label="Export")
plt.plot(year_import.index, year_import.values, marker="o", label="Import")

plt.title("Year-wise Export and Import")
plt.xlabel("Year")
plt.ylabel("Trade Value")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
correlation = pd.DataFrame({
    "Export": year_export,
    "Import": year_import
}).corr()

correlation
```
